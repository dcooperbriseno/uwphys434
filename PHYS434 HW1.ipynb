```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

# this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
```

Chosen Parameters
- new mean = 4.0
- new standard deviation (σ) = 0.02
- sample size = 1,000,000


```python
# loc is the mean/center of the distribution with scale being the standard deviation. The size remains 100,000. 

d = stats.norm.rvs(loc = 4.0, scale = .02, size = 1000000)
```

```python
# After changing the mean, I also changed the bounds of the x-axis on the plot to 
# [3.9,4.1] because I doubled the scale and moved the center down one.
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([3.9,4.1])
x = np.linspace(3.9,4.1,1000)
ax.plot(x,stats.norm.pdf(x,loc = 4., scale = 0.02),linewidth = 8,alpha = 0.7)
plt.show()
```

```python
ax = plt.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([3.9,4.1])
plt.show()
```

## Histogram with PDF Overlay
I adjusted `plt.xlim` to [3.9, 4.1] because my distribution is 
centered at 4.0 with width 0.02, so ±5σ spans roughly [3.9, 4.1].

The theoretical probability density function is:

$$ f(x) = \frac{1}{\sqrt{2\pi(0.02)^2}}e^\frac{-(x-4)^2}{2(0.02)^2} $$

The orange curve plots this PDF using `stats.norm.pdf(x, loc=4., scale=0.02)`. 
Since `density=True` normalizes the histogram, the empirical bars 
should match the theoretical curve.
